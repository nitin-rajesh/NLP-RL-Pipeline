````python
import zipfile
from datasets import load_dataset

OUTPUT_DIR = "biology_data"
NUM_CORES = os.cpu_count()

if not os.path.exists(OUTPUT_DIR):
    print(f"Unzipping {FILE_NAME} to {OUTPUT_DIR}...")
    with zipfile.ZipFile(FILE_NAME, 'r') as zf:
        zf.extractall(OUTPUT_DIR)
    print("✅ Unzipping complete.")
else:
    print(f"Folder {OUTPUT_DIR} already exists. Skipping unzip.")

print(f"Loading data from folder '{OUTPUT_DIR}' with {NUM_CORES} cores...")

dataset = load_dataset(
    "json",
    data_dir=OUTPUT_DIR,
    split="train",
    num_proc=NUM_CORES
)

print(f"✅ Final dataset loaded and parallel processing was enabled successfully!")
````

````python
from vllm import LLM, SamplingParams

llm = LLM(
    model="Qwen/Qwen2.5-3B-Instruct",
    trust_remote_code=True,
    max_model_len=4096,
    gpu_memory_utilization=0.92,
    tensor_parallel_size=1
)

sp = SamplingParams(
    temperature=0.7,
    top_p=0.95,
    max_tokens=900,
    stop_token_ids=[llm.get_tokenizer().eos_token_id]
)

print("vLLM + Qwen2.5-3B loaded and ready!")
````

````python
ENHANCER_SYSTEM = """You are an elite prompt engineer specialized in biology instruction tuning.
Your only job: take a raw biology question and rewrite it into 3 dramatically superior versions that force any LLM to give deeper, more accurate, structured, and reasoning-heavy answers.

Rules:
- Never copy the original prompt.
- Always add strong reasoning (CoT, role-play, formatting, examples, etc.)
- Make 3 clearly different styles:
   1. Olympiad-style: competitive exam, step-by-step, boxed answer
   2. Professor-style: rigorous, evidence-based, real organism examples
   3. Concise-style: ultra-high-signal, perfect for fine-tuning
- Keep 100% of the scientific meaning.

Output ONLY valid JSON:
{
  "enhanced": [
    {"style": "olympiad", "prompt": "...full prompt..."},
    {"style": "professor", "prompt": "...full prompt..."},
    {"style": "concise", "prompt": "...full prompt..."}
  ]
}"""

# Build prompts
prompts = []
for item in dataset:
    user_msg = f"""Original biology question:
"{item['message_1']}"

Topic: {item.get('topic', '')}
Sub-topic: {item.get('sub_topic', '')}

Generate 3 vastly improved versions following the rules above."""

    full = f"<|im_start|>system\n{ENHANCER_SYSTEM}<|im_end|>\n<|im_start|>user\n{user_msg}<|im_end|>\n<|im_start|>assistant\n"
    prompts.append(full)
````

````python
# Generate all at once
print(f"Generating {len(prompts)} questions → {len(prompts)*3:,} enhanced prompts...")
outputs = llm.generate(prompts, sp)
````

````python
import json, re
from tqdm import tqdm
from google.colab import files

output_file = "biology_good_prompts_final.jsonl"
global_id = 1
saved_count = 0

with open(output_file, "w", encoding="utf-8") as f:
    for i, out in tqdm(enumerate(outputs), total=len(outputs)):
        text = out.outputs[0].text.strip()
        item = dataset[i]
        original_prompt = item["message_1"]
        qid = i

        text = text.replace("```json", "").replace("```", "").strip()

        try:
            # Extract JSON
            start = text.find("{")
            end = text.rfind("}") + 1
            if start == -1 or end == 0:
                continue
            json_str = text[start:end]
            data = json.loads(json_str)

            # It outputs "enhanced" + "style" → this is GOOD prompts!
            for enh in data.get("enhanced", []):
                prompt_text = enh.get("prompt", "").strip()
                if not prompt_text:
                    continue

                f.write(json.dumps({
                    "id": global_id,
                    "persona": enh["style"],
                    "qid": qid,
                    "original_prompt": original_prompt,
                    "optimized_prompt": prompt_text,
                    "generator": "Qwen2.5-3B-vLLM"
                }, ensure_ascii=False) + "\n")

                global_id += 1
                saved_count += 1

        except Exception as e:
            continue  # skip rare broken ones

print(f"\nSUCCESS! Saved {saved_count:,} GOOD prompts → {output_file}")
files.download(output_file)
````
